```python
from langchain_openai import ChatOpenAI
from langgraph.prebuilt import create_react_agent
from langgraph_supervisor import create_supervisor

def find_docs(docs: str):
    """Search for documents that contain specific information"""
    return docs

def QAagent(response : str):
    """question answering agent"""
    return response

docs_assistant = create_react_agent(
    model="openai:gpt-4o",
    tools=[find_docs],
    prompt="You are a document search assistant. Your task is to find relevant documents based on user queries.",   
    name="docs_assistant"
)

QA_assistant = create_react_agent(
    model="openai:gpt-4o",
    tools=[QAagent],
    prompt="you are a question answering assistant. Your task is to answer questions based on the documents provided.",
    name="QA_assistant"
)

supervisor = create_supervisor(
    agents=[docs_assistant, QA_assistant],
    model=ChatOpenAI(model="gpt-4o"),
    prompt=(
            "You are a supervisor that coordinates multiple agents to complete tasks. "
    "You will receive a user query and delegate tasks to the appropriate agents. "

        
    )
).compile()

for chunk in supervisor.stream(
    {
        "messages": [
            {
                "role": "user",
                "content": "book a flight from BOS to JFK and a stay at McKittrick Hotel"
            }
        ]
    }
):
    print(chunk)
    print("\n")
```
